```python
root = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/refs/heads/master/'
```

```python
dataframes = []
df_loop = ['seaice', 'planets', 'penguins', 'fmri']
```

```python
last_part = '.csv'
```

```python
for link in df_loop:
    add_links = root + link + last_part
    df = pd.read_csv(add_links)
    dataframes.append(df)
    
```
