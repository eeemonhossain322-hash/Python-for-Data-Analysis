# Project 4:
**Loop Diye Multiple Season/League Load**
***Start with 15 to 20***
### https://football-data.co.uk/mmz4281/1516/E0.csv
### https://football-data.co.uk/mmz4281/1617/E0.csv
### https://football-data.co.uk/mmz4281/1718/E0.csv

```python
root =  'https://football-data.co.uk/mmz4281/'
```

```python
leagues = ['E0', 'E1', 'E2']
frames_pro4 = []
```

```python
for league in leagues:
    for season in range(15, 21):
      df_pro4 = pd.read_csv(root + str(season) + str(season + 1) + '/' + league + '.csv')
      df_pro4.insert(1, 'season', season)
      frames_pro4.append(df_pro4)
df_master = pd.concat(frames_pro4)
```

# Project 5
## *Combining and Organizing Data in the Master Dictionary*

```python
import pandas as pd
```

```python
dict_league_name = {'Premier League': 'E0', 'Championship': 'E1', 'League 1': 'E2'}
print(dict_league_name['Premier League'])
```

```python
dict_concat_data = {}

for league in dict_league_name:
    frames = []
    for season in range(15, 21):
      df_pro4 = pd.read_csv(root + str(season) + str(season + 1) + '/' + dict_league_name[league] + '.csv')
      df_pro4.insert(1, 'season', season)
      frames.append(df_pro4)
    df_master = pd.concat(frames)
    dict_concat_data[league] = df_master
```
